```python
print("Rows before removing null Customer IDs:", len(df))

df = df.dropna(subset=["Customer ID"])

print("Rows after removing null Customer IDs:", len(df))
print("Rows removed:", 1067371 - len(df))
```

```python
returns_df = df[df["Quantity"] < 0].copy()
df = df[df["Quantity"] > 0].copy()

print(len(df))
print(len(returns_df))
```

```python
print("Rows before price filter:", len(df))

df = df[df["Price"] > 0].copy()

print("Rows after price filter:", len(df))
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

print(df["InvoiceDate"].head())
```

```python
df["Revenue"] = df["Quantity"] * df["Price"]

print(df["Revenue"].head())
print("Total revenue: ", round(df["Revenue"].sum(), 2))
```
